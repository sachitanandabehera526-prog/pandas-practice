# Write CSV

```python
import pandas as pd

dis = {"A":[1,2,3,4,5,6],"B":[1,2,3,4,5,6],"C":[1,2,3,4,5,6]}
d = pd.DataFrame(dis)
print(d)
d.to_csv("Test_new.csv")
```

# Read CSV

```python
import pandas as pd
csv_1 = pd.read_csv("Test_new.csv")
csv_1
```

```python
csv_2 = pd.read_csv("Test_new.csv",nrows=3)
csv_2
```

```python
csv_2 = pd.read_csv("Test_new.csv",usecols=["A"])
csv_2
```

```python
csv_2 = pd.read_csv("Test_new.csv",skiprows=[2])
csv_2
```
