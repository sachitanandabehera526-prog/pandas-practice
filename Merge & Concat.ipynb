# Merge

```python
import pandas as pd
var1 = pd.DataFrame({"A":[1,2,3,4,5], "B":[11,12,13,14,15]})
var2 = pd.DataFrame({"A":[1,2,3,4,6], "C":[21,22,23,24,25]})
pd.merge(var1,var2,on="A")
```

```python
pd.merge(var1,var2,how="left")
```

```python
pd.merge(var1,var2,how="right")
```

```python
pd.merge(var1,var2,how="outer")
```

```python
pd.merge(var1,var2,left_index=True,right_index=True)
```

# Concat

```python
sr1 = pd.Series([1,2,3,4,5])
sr2 = pd.Series([12,24,35,48,50])
pd.concat([sr1,sr2])
```

```python
d1 = pd.DataFrame({"A":[1,2,3,4,5], "B":[11,12,13,14,15]})
d2 = pd.DataFrame({"A":[1,2,3,4,6], "B":[21,22,23,24,25]})
pd.concat([d1,d2])
```

```python
pd.concat([d1,d2],axis=1,join="outer")
```

```python
pd.concat([d1,d2],axis=0,join="inner")
```

```python
pd.concat([d1,d2],axis=1,keys=["d1","d2"])
```

```python
pd.concat([d1,d2],axis=0,keys=["d1","d2"])
```
